```python
html_c = data.text
```

```python
print(data.content[0:100])
```

```python
# We can also introduce using time.sleep() in scraping to avoid hitting the server too fast.
```

```python
from bs4 import BeautifulSoup
```

```python
soup = BeautifulSoup(html_c,'lxml')
```

```python
# soup
```

```python
# Methods and attributes
```

```python
# Country name : H3  class : country-name
```

## Methods : 
### find : 
- find()
- find_all()
- find_parent()
- find_next()
- find_previous()
### css selectors : 
- select()
- select_one()
### attributes : 
- get()
- .attrs

```python
soup.find('h1')
```

```python
country_name = soup.find_all('h3')
```

```python
type(country_name) # list
```

```python
for h3 in country_name:
    print(h3.get_text(strip=True))
```

```python
# access population
```

```python
all_countries = [] 
for h3 in country_name:
    name = h3.get_text(strip=True)
    population = h3.find_next('div').select('span.country-population')[0].get_text()
    all_countries.append([name,population])
```

```python
all_countries
```

```python
import pandas as pd
```

```python
cp = pd.DataFrame(all_countries, columns=['Country Name', 'Population'])
```

```python
cp
```

```python
cp.to_csv('PopulationData',index=False)
```
