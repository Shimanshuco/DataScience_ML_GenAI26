# Pokémon Explorer
<a href="https://pokeapi.co/"> Visit Pokémon API </a>

```python
import pandas as pd
import requests
```

```python
url = 'https://pokeapi.co/api/v2/pokemon?limit=100'
```

```python
res = requests.get(url)
res
```

```python
pokemon = res.json()
pokemon
```

```python
print(len(pokemon["results"]))
```

```python
df = pd.DataFrame(pokemon["results"])

df
```

```python
data = []

for item in pokemon["results"]:
    detail = requests.get(item["url"]).json()

    data.append({
        "ID": detail["id"],
        "Name": detail["name"].title(),
        "Height": detail["height"],
        "Weight": detail["weight"],
        "Type": ", ".join(t["type"]["name"] for t in detail["types"]),
        "Abilities": ", ".join(a["ability"]["name"] for a in detail["abilities"]),
        "Image": detail["sprites"]["front_default"]
    })

df = pd.DataFrame(data)

df
```

```python
df.to_csv("pokemon.csv", index=False)
```

```python
from IPython.display import display, Image
```

```python
def get_pokemon_by_id(pokemon_id):
    result = df[df["ID"] == pokemon_id]

    if result.empty:
        print("Pokémon not found!")
        return

    display(result.drop(columns=["Image"]))

    print("\nPokémon Image:\n")
    display(Image(url=result.iloc[0]["Image"]))
```

```python
get_pokemon_by_id(25)
```

```python
get_pokemon_by_id(1)
```
