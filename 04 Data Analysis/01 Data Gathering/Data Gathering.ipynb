# Importing a CSV File

## What is a CSV File?

**CSV** stands for **Comma-Separated Values**.

- It is one of the most commonly used file formats for storing tabular data.
- Data is stored in rows and columns.
- Each value in a row is separated by a comma (`,`).
- It is widely used because it is simple, lightweight, and supported by almost every data analysis tool.

### Example (students.csv)

```csv
ID,Name,Age,Marks
101,Amit,20,85
102,Priya,21,90
103,Rahul,19,78
```

---

# Syntax to Import a CSV File

```python
import pandas as pd

df = pd.read_csv("students.csv")
```

````python
# Import pandas library
import pandas as pd

# Create DataFrame with separate date columns
data = {
    "day": [15, 20, 5, 10],
    "month": [8, 9, 10, 11],
    "year": [2024, 2024, 2025, 2025]
}

df = pd.DataFrame(data)

# Display original DataFrame
print("Original DataFrame:")
print(df)


# Combine day, month, and year columns into one date column
df["date"] = pd.to_datetime(
    df[["year", "month", "day"]]
)

# Display updated DataFrame
print("\nDataFrame after combining date columns:")
print(df)


# Check data type of date column
print("\nData Types:")
print(df.dtypes)

print("\nExtracted Date Information:")
print(df)
print(df.info())
````

### Excel

# Reading Excel Files

Pandas provides the `read_excel()` function to import data from Excel files (`.xlsx` or `.xls`) into a DataFrame.

---

# Basic Syntax

```python
import pandas as pd

df = pd.read_excel("filename.xlsx")
```

### Explanation:

- `pd` → Pandas library alias
- `read_excel()` → Function used to read Excel files
- `"filename.xlsx"` → Excel file path
- `df` → Stores the data as a DataFrame

---

# Example Excel File (`students.xlsx`)

| ID | Name | Age | Marks |
|----|------|-----|-------|
| 101 | Amit | 21 | 85 |
| 102 | Priya | 22 | 90 |
| 103 | Rahul | 20 | 78 |

---

# 1. Read Excel File

```python
import pandas as pd

df = pd.read_excel("students.xlsx")

print(df)
```

Output:

```
    ID   Name  Age  Marks
0  101   Amit   21     85
1  102  Priya   22     90
2  103  Rahul   20     78
```

---

# 2. Read Specific Sheet

Excel files can contain multiple sheets.

Syntax:

```python
df = pd.read_excel(
    "students.xlsx",
    sheet_name="Sheet1"
)
```

Example:

```python
df = pd.read_excel(
    "students.xlsx",
    sheet_name="Student_Data"
)
```

---

# 3. Read Sheet Using Sheet Index

Sheets can also be selected by their position.

```python
df = pd.read_excel(
    "students.xlsx",
    sheet_name=0
)
```

Meaning:

- `0` → First sheet
- `1` → Second sheet
- `2` → Third sheet

---

# 4. Read All Sheets

Use:

```python
df = pd.read_excel(
    "students.xlsx",
    sheet_name=None
)
```

Output:

```python
{
'Sheet1': DataFrame,
'Sheet2': DataFrame
}
```

It returns a dictionary of DataFrames.

---

# 5. Read Specific Columns

Use `usecols`.

```python
df = pd.read_excel(
    "students.xlsx",
    usecols=["Name", "Marks"]
)
```

Output:

| Name | Marks |
|------|-------|
| Amit | 85 |
| Priya | 90 |

---

# 6. Read Specific Rows

Use `nrows`.

```python
df = pd.read_excel(
    "students.xlsx",
    nrows=5
)
```

Reads only the first 5 rows.

---

# 7. Skip Rows While Reading

Use `skiprows`.

```python
df = pd.read_excel(
    "students.xlsx",
    skiprows=2
)
```

Skips the first 2 rows.

---

# 8. Reading Excel Without Header

If the first row is not a column name:

```python
df = pd.read_excel(
    "students.xlsx",
    header=None
)
```

Pandas will assign default column numbers:

```
0   1      2
Amit 21   85
```

---

# 9. Assign Custom Column Names

```python
df = pd.read_excel(
    "students.xlsx",
    header=None,
    names=["Name", "Age", "Marks"]
)
```

---

# 10. Set Index Column

```python
df = pd.read_excel(
    "students.xlsx",
    index_col="ID"

```

```

ID
```

# 11. Read Date Columns

If Excel contains dates:

```python
df = pd.read_excel(
    "students.xlsx",
    parse_dates=["Joining_Date"]
)
```

Pandas converts the column into datetime format.

---

# 12. Specify Data Types

Use `dtype`.

```python
df = pd.read_excel(
    "students.xlsx",
    dtype={"ID": str}
)
```

Example:

Before:

```
101
```

After:

```
"101"
```

# 13. Handle Missing Values

```python
df = pd.read_excel(
    "students.xlsx",
    na_values=["NA", "Missing", "-"]
```

These values will be treated as `NaN`.

## What is a Text File?

A **text file (`.txt`)** stores data in plain text format.

Examples:

```
students.txt
employees.txt
sales.txt
```

Text files can store data separated by:

- Comma (`,`)
- Tab (`\t`)
- Space
- Semicolon (`;`)
- Other delimiters

Pandas can read text files using the `read_csv()` function because a text file is also a type of delimited file.

---

# Reading JSON Files

## What is a JSON File?

**JSON (JavaScript Object Notation)** is a lightweight format used to store and exchange data.

JSON data is stored in:

- Key-value pairs
- Arrays
- Nested objects

It is commonly used in:
- APIs
- Web applications
- Data exchange between systems

File extension:

```
.json
```

# JSON Orientation (`orient` Parameter)

JSON can store data in different formats.

The `orient` parameter tells Pandas how JSON data is structured.

---

## 1. Records Format

Example:

```json
[
 {"Name":"Amit","Age":21},
 {"Name":"Priya","Age":22}
]
```

Read:

```python
df = pd.read_json(
    "students.json",
    orient="records"
)
```

---

## 2. Columns Format

Example:

```json
{
 "Name":{
    "0":"Amit",
    "1":"Priya"
 },
 "Age":{
    "0":21,
    "1":22
 }
}
```

Read:

```python
df = pd.read_json(
    "students.json",
    orient="columns"
)
```

---

## 3. Index Format

```python
df = pd.read_json(
    "students.json",
    orient="index"
)
```

# Import Data from SQL Database 

## What is SQL Data Import?

In Data Analysis, data is often stored in databases like:

- MySQL
- PostgreSQL
- SQLite
- SQL Server
- Oracle

Pandas can directly read data from SQL databases and convert it into a **DataFrame**.

---

# Required Libraries

```python
import pandas as pd
import sqlite3
```

`sqlite3` is a built-in Python library used to connect with SQLite databases.

For other databases, install connectors:

```bash
pip install mysql-connector-python
```

or

```bash
pip install psycopg2
```

---

# Method 1: Read SQL Table Using SQLite

## Step 1: Create Database Connection

```python
import sqlite3

connection = sqlite3.connect("students.db")
```

---

## Step 2: Read SQL Table

### Syntax:

```python
pd.read_sql(
    "SQL Query",
    connection
)
```

Example:

```python
df = pd.read_sql(
    "SELECT * FROM students",
    connection
)

print(df)
```

---

# Complete Example

```python
import pandas as pd
import sqlite3

# Create connection
connection = sqlite3.connect("students.db")

# Read data from SQL table
df = pd.read_sql(
    "SELECT * FROM students",
    connection
)

# Display DataFrame
print(df)

# Close connection
connection.close()
```

---

# Method 2: Read SQL Query

Instead of reading the complete table, we can use SQL queries.

Example:

```python
df = pd.read_sql(
    """
    SELECT Name, Marks
    FROM students
    WHERE Marks > 80
    """,
    connection
)

print(df)
```

Output:

```
    Name   Marks
0   Amit      85
1  Priya      90
```

---
