
- ## `astype`

## Function
`astype(dtype, copy=True, errors='raise')`

## What it does
- Converts the data type (`dtype`) of a pandas Series or DataFrame.
- Allows changing one or multiple columns to a specified data type.
- Commonly used for type conversion before performing data analysis or computations.

## Important Points
- Accepts a single data type or a dictionary mapping column names to their target data types.
- By default, a copy of the object is returned (`copy=True`).
- The `errors` parameter controls conversion behavior:
  - `'raise'` (default): Raises an exception if conversion fails.
  - `'ignore'`: Returns the original object if conversion cannot be performed.
- Incompatible conversions may raise errors unless `errors='ignore'` is used.
- Can convert data to NumPy data types as well as pandas extension data types (such as nullable integer, string, or boolean types).
- Missing values may affect conversions depending on the target data type.
- Returns a Series or DataFrame with the specified data type(s).

## `between`

## Function
`between(left, right, inclusive='both')`

## What it does
- Checks whether each value in a pandas Series lies within a specified range.
- Returns a boolean Series indicating whether each element falls between the lower and upper bounds.

## Important Points
- Available only for **Series** objects.
- By default, both boundary values are included (`inclusive='both'`).
- The `inclusive` parameter supports:
  - `'both'` – Includes both boundaries.
  - `'left'` – Includes only the left boundary.
  - `'right'` – Includes only the right boundary.
  - `'neither'` – Excludes both boundaries.
- Works with numeric values, dates, timestamps, and other comparable data types.
- Missing values (`NaN`) always return `False`.
- Returns a boolean Series of the same length as the original Series.

````python
subs[subs.between(50,100)]
````

## `clip`

## Function
`clip(lower=None, upper=None, axis=None, inplace=False, **kwargs)`

## What it does
- Limits values in a Series or DataFrame to specified lower and/or upper boundaries.
- Values smaller than the lower limit are replaced with the lower limit.
- Values larger than the upper limit are replaced with the upper limit.

## Important Points
- Can be applied to both **Series** and **DataFrame** objects.
- The `lower` and `upper` parameters are optional; either one can be specified independently.
- Supports scalar values as well as array-like thresholds for element-wise clipping.
- The `axis` parameter is mainly used when applying array-like thresholds to a DataFrame.
- By default, returns a new object. Use `inplace=True` to modify the original object.
- Missing values (`NaN`) remain unchanged during clipping.
- Returns a Series or DataFrame with values constrained within the specified range.

````python
subs.clip(100,200)
````

## `drop_duplicates`

## Function
`drop_duplicates(subset=None, keep='first', inplace=False, ignore_index=False)`

## What it does
- Removes duplicate rows from a DataFrame or duplicate values from a Series.
- Retains only the desired occurrence of each duplicate based on the specified criteria.

## Important Points
- Works with both **Series** and **DataFrame** objects.
- By default, duplicate rows are identified using all columns.
- The `subset` parameter allows checking duplicates based on one or more specific columns.
- The `keep` parameter determines which duplicate to retain:
  - `'first'` – Keeps the first occurrence (default).
  - `'last'` – Keeps the last occurrence.
  - `False` – Removes all duplicate occurrences.
- By default, returns a new object. Use `inplace=True` to modify the original object.
- The `ignore_index=True` parameter resets the index in the returned object.
- Missing values (`NaN`) are considered duplicates and are handled according to the `keep` parameter.
- Returns a Series or DataFrame with duplicate entries removed.

````python
temp = pd.Series([1,1,2,2,3,4,4])
temp
````

````python
temp.drop_duplicates() # Keeps first occurrences
````

````python
temp = pd.Series([1,1,2,2,3,4,4])
temp
````

````python
temp.drop_duplicates(keep='last') #Keeps last occurrences
````

## `duplicated`

## Function
`duplicated(subset=None, keep='first')`

## What it does
- Identifies duplicate rows in a DataFrame or duplicate values in a Series.
- Returns a boolean mask indicating whether each row or value is a duplicate.

## Important Points
- Works with both **Series** and **DataFrame** objects.
- By default, duplicate rows are identified using all columns.
- The `subset` parameter allows checking duplicates based on one or more specific columns.
- The `keep` parameter determines which occurrence is marked as unique:
  - `'first'` – Marks all duplicates except the first occurrence as `True` (default).
  - `'last'` – Marks all duplicates except the last occurrence as `True`.
  - `False` – Marks all duplicate occurrences as `True`.
- Missing values (`NaN`) are considered duplicates and are handled according to the `keep` parameter.
- Returns a boolean Series that can be used for filtering or identifying duplicate data.

````python
temp = pd.Series([1,1,2,2,3,4,4])
temp.duplicated()
````

````python
temp.duplicated().sum()
````

## `isnull`

## Function
`isnull()`

## What it does
- Detects missing (null) values in a Series or DataFrame.
- Returns a boolean object where missing values are marked as `True` and non-missing values as `False`.

## Important Points
- Works with both **Series** and **DataFrame** objects.
- Identifies values such as `NaN`, `None`, and `NaT` as missing.
- Returns a boolean Series or DataFrame with the same shape as the original object.
- Commonly used for filtering, counting, or locating missing values.
- `isnull()` is an alias of `isna()`; both functions produce identical results.
- Does not modify the original data.

````python
s = pd.Series([10, 20, np.nan, 40, None, 60, pd.NA])
s
````

````python
print(s.size)
print(s.count())

# size = total elements (including nulls)
# count() = only valid (non-null) elements
````

## `dropna`

## Function
`dropna(axis=0, how='any', thresh=None, subset=None, inplace=False)`

## What it does
- Removes missing values (`NaN`, `None`, `pd.NA`) from a Series or DataFrame.
- Can drop rows or columns depending on the specified axis and conditions.

## Important Points
- Works on both **Series and DataFrame** objects.
- The `axis` parameter controls removal direction:
  - `axis=0` → drops rows (default)
  - `axis=1` → drops columns
- The `how` parameter defines dropping condition:
  - `'any'` → drops if any missing value is present (default)
  - `'all'` → drops only if all values are missing
- The `thresh` parameter keeps rows/columns with at least a minimum number of non-null values.
- The `subset` parameter allows checking missing values only in selected columns.
- By default, returns a new object; `inplace=True` modifies the original data.
- Returns a cleaned Series or DataFrame with missing values removed.

````python
s.dropna()
````

## `fillna`

## Function
`fillna(value=None, method=None, axis=None, inplace=False, limit=None, downcast=None)`

## What it does
- Replaces missing values (`NaN`, `None`, `pd.NA`) in a Series or DataFrame with specified values or methods.
- Helps in handling incomplete data by filling gaps.

## Important Points
- Works on both **Series and DataFrame** objects.
- The `value` parameter can be a scalar, dictionary, Series, or DataFrame used to fill missing values.
- The `method` parameter allows filling using propagation:
  - `'ffill'` / `'pad'` → forward fill (uses previous valid value)
  - `'bfill'` / `'backfill'` → backward fill (uses next valid value)
- The `axis` parameter defines direction of filling (row-wise or column-wise).
- The `limit` parameter restricts the number of consecutive NaNs to fill.
- By default, returns a new object; `inplace=True` modifies the original data.
- Does not modify non-missing values.
- Returns a Series or DataFrame with missing values replaced.

````python
s = pd.Series([10, 20, np.nan, 40, None, 60, pd.NA])
s
````

````python
s.fillna(0)
````

````python
s.fillna(s.mean())
````

## `isin`

## Function
`isin(values)`

## What it does
- Checks whether each element in a Series or DataFrame is present in a given set of values.
- Returns a boolean object indicating membership (True if present, False otherwise).

## Important Points
- Works with both **Series and DataFrame** objects.
- Commonly used for filtering data based on multiple values.
- The `values` parameter can be a list, set, Series, or dictionary-like object.
- For DataFrames, membership is checked element-wise.
- Returns a boolean Series or DataFrame with the same shape as input.
- Missing values (`NaN`) always return `False`.
- Useful for creating conditional filters in data analysis.

````python
subs[(subs == 49 ) | (subs == 99)]
````

````python
subs.isin([49,99])
````

````python
subs[subs.isin([49,99])]
````

## `apply`

## Function
`apply(func, axis=0, raw=False, result_type=None, args=(), **kwargs)`

## What it does
- Applies a custom function along an axis of a Series or DataFrame.
- Used to perform row-wise or column-wise operations in DataFrames, or element-wise transformations in Series.

## Important Points
- Works on both **Series and DataFrame** objects.
- In DataFrames:
  - `axis=0` → function applied column-wise
  - `axis=1` → function applied row-wise
- In Series:
  - Function is applied element-wise by default.
- Allows passing custom Python functions, including lambda functions.
- The `args` and `kwargs` parameters allow passing additional arguments to the function.
- The `result_type` parameter controls the output structure for DataFrames (when applicable).
- Can return Series, DataFrame, or scalar depending on the function used.
- More flexible but often slower than vectorized pandas operations.

````python
boll.apply(lambda x:x.split()[0].upper())
````

````python
subs.mean()
````

````python
subs.apply(lambda x:'Good Day :-)' if x >subs.mean() else 'Bad Day :-(')
````

## `copy`

## Function
`copy(deep=True)`

## What it does
- Creates a copy of a pandas Series or DataFrame.
- Used to duplicate data so that changes in the new object do not affect the original object.

## Important Points
- Works on both **Series and DataFrame** objects.
- By default, `deep=True`:
  - Creates a **deep copy**, meaning both data and structure are fully copied.
- If `deep=False`:
  - Creates a **shallow copy**, where only the reference is copied (some changes may reflect in original data).
- Useful to avoid **SettingWithCopyWarning** in pandas.
- Ensures safe manipulation of data without modifying the original dataset unintentionally.
- Returns a completely new Series or DataFrame object.
