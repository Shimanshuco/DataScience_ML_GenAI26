### Series Methods

## `head()`

### Purpose
Returns the first `n` rows/elements of a Series.

### Syntax
```python
series.head(n)
```

### Parameters
- `n` *(optional)*: Number of rows to return.
- Default: `5`

### Returns
- Series containing the first `n` elements.

### Common Usage
```python
series.head()      # First 5 elements
series.head(10)    # First 10 elements
series.head(1)     # First element
```

### Important Points
- Used for quick data inspection.
- Does not modify the original Series.
- Returns a new Series.
- If `n` is larger than the Series length, the entire Series is returned.

````python
s.head()
````

````python
movie.head(10)
````

## `tail()`

### Purpose
Returns the last `n` rows/elements of a Series.

### Syntax
```python
series.tail(n)
```

### Parameters
- `n` *(optional)*: Number of rows to return.
- Default: `5`

### Returns
- Series containing the last `n` elements.

### Common Usage
```python
series.tail()      # Last 5 elements
series.tail(10)    # Last 10 elements
series.tail(1)     # Last element
```

### Important Points
- Used for inspecting the end of the data.
- Does not modify the original Series.
- Returns a new Series.
- If `n` is larger than the Series length, the entire Series is returned.
- Supports negative values:
  ```python
  series.tail(-3)
  ```
  Returns all rows except the first 3.

````python
s.tail()
````

````python
s.tail(10)
````

````python
s.tail(-3)
````

## `sample()`

### Purpose
Returns a random sample of elements from a Series.

### Syntax
```python
series.sample(n=None, frac=None, replace=False, random_state=None)
```

### Parameters
- `n` *(optional)*: Number of items to return.
- `frac` *(optional)*: Fraction of total items to return (0 to 1).
- `replace` *(optional)*: Whether sampling is with replacement (`True/False`).
- `random_state` *(optional)*: Seed for reproducibility.

### Returns
- Randomly selected elements from the Series.

### Common Usage
```python
series.sample()          # 1 random element
series.sample(5)         # 5 random elements
series.sample(frac=0.2)  # 20% of data
```

### Important Points
- Used for random inspection of data.
- Cannot use both `n` and `frac` together.
- Results change every run unless `random_state` is set.
- Does not modify original Series.

````python
s.sample()
````

````python
movie.sample()
````

````python
movie.sample(frac=0.2)
````

## `value_counts()`

### Purpose
Counts the frequency of each unique value in a Series.

### Syntax
```python
series.value_counts(normalize=False, sort=True, ascending=False, dropna=True)
```

### Parameters
- `normalize` *(bool)*: If True, returns proportions instead of counts.
- `sort` *(bool)*: Sort results by frequency.
- `ascending` *(bool)*: Sort order (default False = descending).
- `dropna` *(bool)*: Whether to exclude NaN values.

### Returns
- Series with unique values as index and counts as values.

### Common Usage
```python
series.value_counts()              # frequency count
series.value_counts(normalize=True) # percentage distribution
series.value_counts(dropna=False)   # include NaN
```

### Important Points
- Most frequently used for categorical data analysis.
- Output is sorted by default (highest frequency first).
- Returns a Series sorted by count, not by original order.
- NaN is ignored by default unless `dropna=False`.

````python
s.value_counts()
````

````python
movie.value_counts()
````

## `sort_values()`

### Purpose
Sorts a Series by its values.

### Syntax
```python
series.sort_values(ascending=True, inplace=False, na_position='last', ignore_index=False)
```

### Parameters
- `ascending` *(bool)*: Sort order  
  - `True` → ascending (default)  
  - `False` → descending  
- `inplace` *(bool)*: Modify original Series or not  
- `na_position` *(str)*: Position of NaN (`'first'` or `'last'`)  
- `ignore_index` *(bool)*: Reset index after sorting

### Returns
- Sorted Series

### Common Usage
```python
series.sort_values()                 # ascending
series.sort_values(ascending=False)  # descending
series.sort_values(inplace=True)     # modify original
```

### Important Points
- Sorts based on values, not index.
- Index remains unchanged unless `ignore_index=True`.
- NaN values are placed at the end by default.
- Does not group duplicates; only orders them.

````python
s.sort_values()
````

````python
movie.sort_values()
````

````python
# Method chaining
print(s.sort_values())
print()
print(s.sort_values(ascending=False))
print()
print(s.sort_values(ascending=False).head(1))
print()
print(s.sort_values(ascending=False).head(1).values)
print()
print(s.sort_values(ascending=False).head(1).values[0])
````

## `sort_index()`

### Purpose
Sorts a Series based on its index labels.

### Syntax
```python
series.sort_index(ascending=True, inplace=False, na_position='last')
```

### Parameters
- `ascending` *(bool)*: Sort order of index  
  - `True` → ascending (default)  
  - `False` → descending  
- `inplace` *(bool)*: Modify original Series or not  
- `na_position` *(str)*: Position of NaN in index (`'first'` or `'last'`)

### Returns
- Series sorted by index

### Common Usage
```python
series.sort_index()                 # sort index ascending
series.sort_index(ascending=False)  # sort index descending
series.sort_index(inplace=True)     # modify original
```

### Important Points
- Sorting is based on index labels, not values.
- Useful when index is meaningful (dates, names, IDs).
- Does not change values, only rearranges order.
- Works with both numeric and string indices.

````python
s.sort_index() 
````

````python
movie.sort_index() 
````

````python
movie.sort_index(ascending=False)
````

````python
# Original List remain Same
movie
````

````python
# To change the original list also 
movie.sort_index(inplace=True)
````

````python
movie # Change Successfull
````

### Series Maths Method

## `count()`

### Purpose
Counts the number of **non-null (non-NaN)** values in a Series.

### Syntax
```python
series.count()
```

### Parameters
- None

### Returns
- Integer count of non-null values

### Common Usage
```python
series.count()
```

### Important Points
- Ignores `NaN` values.
- Does NOT count missing values.
- Different from `len()`:
  - `count()` → non-null values only  
  - `len()` → total values (including NaN)
- Useful for checking data completeness.

````python
s.count()
````

````python
movie.count()
````

## `sum()`

### Purpose
Returns the sum of all values in a Series.

### Syntax
```python
series.sum(skipna=True, min_count=0)
```

### Parameters
- `skipna` *(bool)*: Whether to ignore NaN values (default True)
- `min_count` *(int)*: Minimum number of valid values required to return a result

### Returns
- Sum of Series values

### Common Usage
```python
series.sum()
```

### Important Points
- Works mainly on numeric data.
- Ignores `NaN` values by default.
- For empty or all-NaN Series, result may be 0 or NaN depending on `min_count`.
- For string Series, it concatenates strings.

````python
s.sum()
````

## `product()`

### Purpose
Returns the product (multiplication) of all values in a Series.

### Syntax
```python
series.product(skipna=True, min_count=0)
```

### Parameters
- `skipna` *(bool)*: Ignore NaN values (default True)
- `min_count` *(int)*: Minimum number of valid values required to return a result

### Returns
- Product of all values in the Series

### Common Usage
```python
series.product()
```

### Important Points
- Works mainly on numeric data.
- Ignores `NaN` values by default.
- If any value is `0`, result becomes `0`.
- If Series is empty or all NaN, result depends on `min_count`.
- Can grow very large quickly (overflow risk).

## `std()`

### Purpose
Returns the standard deviation of values in a Series.

### Syntax
```python
series.std(skipna=True, ddof=1)
```

### Parameters
- `skipna` *(bool)*: Ignore NaN values (default True)
- `ddof` *(int)*: Delta degrees of freedom (default 1 for sample standard deviation)

### Returns
- Standard deviation (spread of data around mean)

### Common Usage
```python
series.std()
```

### Important Points
- Works only with numeric data.
- Measures how spread out the values are.
- Higher value → more variability in data.
- By default uses **sample standard deviation (ddof=1)**.
- Ignores NaN values by default.

````python
s.std()
````

## `var()`

### Purpose
Returns the variance of values in a Series.

### Syntax
```python
series.var(skipna=True, ddof=1)
```

### Parameters
- `skipna` *(bool)*: Ignore NaN values (default True)
- `ddof` *(int)*: Delta degrees of freedom (default 1 for sample variance)

### Returns
- Variance of the Series values

### Common Usage
```python
series.var()
```

### Important Points
- Works only with numeric data.
- Measures spread of data around the mean.
- Square of standard deviation: `var() = std()²`.
- By default uses **sample variance (ddof=1)**.
- Ignores NaN values by default.

````python
s.var()
````

## `min()` / `max()`

### Purpose
Returns the smallest (`min`) and largest (`max`) values in a Series.

### Syntax
```python
series.min(skipna=True)
series.max(skipna=True)
```

### Parameters
- `skipna` *(bool)*: Ignore NaN values (default True)

### Returns
- `min()` → smallest value in the Series  
- `max()` → largest value in the Series  

### Common Usage
```python
series.min()
series.max()
```

### Important Points
- Works with numeric and comparable data (strings also supported).
- Ignores `NaN` values by default.
- For strings:
  - `min()` → lexicographically smallest
  - `max()` → lexicographically largest
- Useful for range detection and data validation.

````python
s.min()
````

````python
s.max()
````

## `describe()`

### Purpose
Generates a **summary statistics report** of a Series.

### Syntax
```python
series.describe(percentiles=None, include=None, exclude=None)
```

### Parameters
- `percentiles` *(list-like, optional)*: Custom percentiles (default: 25%, 50%, 75%)
- `include` *(optional)*: Data types to include
- `exclude` *(optional)*: Data types to exclude

### Returns
- A Series containing summary statistics

---

## Output depends on data type

### 1. Numeric Series
```python
series.describe()
```

Returns:
- `count`
- `mean`
- `std`
- `min`
- `25%`
- `50%` (median)
- `75%`
- `max`

---

### 2. Categorical / Object Series
Returns:
- `count`
- `unique`
- `top` (most frequent value)
- `freq` (frequency of top value)

---

### Common Usage
```python
series.describe()
```

---

### Important Points
- Quick overview of dataset distribution.
- Works differently for numeric vs categorical data.
- Ignores NaN values by default.
- Very useful for initial data analysis (EDA).

````python
s.describe()
````
