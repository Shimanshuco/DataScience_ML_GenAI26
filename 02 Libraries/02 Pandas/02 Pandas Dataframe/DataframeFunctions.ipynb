## `value_counts`

## Function
`value_counts(subset=None, normalize=False, sort=True, ascending=False, dropna=True)`

## What it does
- Counts the number of occurrences of each unique value in a Series or DataFrame column.
- Returns a Series with unique values as the index and their frequencies as values.

## Important Points
- Works mainly on a **Series** (or a single DataFrame column).
- By default, results are sorted in descending order of frequency.
- The `normalize` parameter:
  - `False` → returns counts (default)
  - `True` → returns relative frequencies (proportions)
- The `dropna` parameter controls whether missing values are included in the result.
- Returns a Series sorted by frequency unless `sort=False` is set.
- Commonly used for exploratory data analysis (EDA) to understand distribution of values.

## `sort_values`

## Function
`sort_values(by, axis=0, ascending=True, inplace=False, kind='quicksort', na_position='last', ignore_index=False, key=None)`

## What it does
- Sorts a DataFrame by one or more columns.
- Returns a new DataFrame with rows arranged according to the specified column(s).
- Can sort in ascending or descending order.

## Important Points
- The `by` parameter specifies the column(s) to sort by (required).
- `ascending=True` sorts in ascending order, while `False` sorts in descending order.
- `inplace=True` modifies the original DataFrame instead of returning a new one.
- `na_position` controls the placement of missing values (`'last'` by default or `'first'`).
- Multiple columns can be used for sorting by passing a list to `by`.
- `ignore_index=True` resets the index in the sorted DataFrame.
- Supports custom sorting using the `key` parameter.
- Commonly used in data cleaning, ranking, and preparing data for analysis.

## `rank`

## Function
`rank(axis=0, method='average', numeric_only=False, na_option='keep', ascending=True, pct=False)`

## What it does
- Assigns ranks to values in a Series or DataFrame.
- Higher or lower values receive ranks based on the sorting order.
- Handles duplicate values using different ranking methods.

## Important Points
- Works on both **Series** and **DataFrame**.
- By default, duplicate values receive the **average** of their ranks (`method='average'`).
- Common `method` options:
  - `'average'` → Average rank for duplicates (default)
  - `'min'` → Lowest rank among duplicates
  - `'max'` → Highest rank among duplicates
  - `'first'` → Ranks assigned based on order of appearance
  - `'dense'` → Like `'min'`, but without gaps in rank numbers
- `ascending=True` ranks from smallest to largest, while `False` ranks from largest to smallest.
- `na_option` controls how missing values are ranked:
  - `'keep'` → Leaves NaN as NaN (default)
  - `'top'` → Places NaN at the top
  - `'bottom'` → Places NaN at the bottom
- `pct=True` returns percentage ranks instead of integer ranks.
- Commonly used for ranking students, salaries, scores, sales, and leaderboard data.

## `sort_index`

## Function
`sort_index(axis=0, level=None, ascending=True, inplace=False, kind='quicksort', na_position='last', sort_remaining=True, ignore_index=False, key=None)`

## What it does
- Sorts a DataFrame or Series by its index (row labels or column labels).
- Returns a new object with the index arranged in sorted order.
- Can sort either rows (`axis=0`) or columns (`axis=1`).

## Important Points
- Works on both **Series** and **DataFrame**.
- By default, sorts **row indices** in ascending order (`axis=0`).
- Use `axis=1` to sort **column names** alphabetically.
- `ascending=True` sorts in ascending order, while `False` sorts in descending order.
- `inplace=True` modifies the original object instead of returning a new one.
- `ignore_index=True` resets the index after sorting.
- Supports custom sorting using the `key` parameter.
- Useful when the index is unordered or after operations like concatenation, merging, or filtering.
- Unlike `sort_values()`, `sort_index()` sorts based on **index labels**, not the data values.

```python
movies.sort_index(ascending=False)
```

## `set_index`

## Function
`set_index(keys, drop=True, append=False, inplace=False, verify_integrity=False)`

## What it does
- Sets one or more existing DataFrame columns as the index.
- Returns a new DataFrame with the specified column(s) as the row index.
- Can create a single-level or MultiIndex.

## Important Points
- Works only on a **DataFrame**.
- The `keys` parameter specifies the column(s) to use as the new index.
- `drop=True` removes the column(s) used as the index from the DataFrame (default).
- `drop=False` keeps the column(s) even after setting them as the index.
- `append=True` adds the new index to the existing index, creating a **MultiIndex**.
- `inplace=True` modifies the original DataFrame instead of returning a new one.
- `verify_integrity=True` checks for duplicate index values and raises an error if duplicates exist.
- Commonly used to improve data organization and enable faster label-based indexing with `loc[]`.
- Use `reset_index()` to convert the index back into regular column(s).

## `reset_index`

`reset_index(level=None, drop=False, inplace=False, col_level=0, col_fill='', allow_duplicates=False, names=None)`

## What it does
- Resets the DataFrame index to the default integer index.
- Moves the current index into one or more columns by default.
- Returns a new DataFrame with a fresh sequential index.

## Important Points
- Works mainly on a **DataFrame** (also available for **Series**).
- By default, the existing index is added as a column (`drop=False`).
- `drop=True` removes the current index instead of adding it as a column.
- `inplace=True` modifies the original DataFrame instead of returning a new one.
- `level` is used to reset specific level(s) of a **MultiIndex**.
- `names` allows assigning custom names to the inserted index column(s).
- Commonly used after filtering, sorting, concatenation, or grouping to restore a clean integer index.
- Often used together with `set_index()` to switch between index-based and column-based representations.

- Useful for checking the variety or uniqueness of data.

## `fillna`

## Function
`fillna(value=None, method=None, axis=None, inplace=False, limit=None)`

## What it does
- Replaces missing values (`NaN`) in a Series or DataFrame.
- Fills null values with a specified value or by propagating existing values.
- Returns a new object with missing values filled unless `inplace=True`.

## Important Points
- Works on both **Series** and **DataFrame**.
- `value` can be a scalar, dictionary, Series, or DataFrame.
- `method='ffill'` fills missing values using the previous valid value (forward fill).
- `method='bfill'` fills missing values using the next valid value (backward fill).
- `inplace=True` modifies the original object instead of returning a new one.
- `axis=0` fills values down each column, while `axis=1` fills across each row (when applicable).
- `limit` specifies the maximum number of consecutive `NaN` values to fill.
- Different columns can be filled with different values using a dictionary.
- Commonly used during data cleaning to handle missing values before analysis or machine learning.

```python
students['Name'].fillna('unknown') #series
```

```python
#dataframe
students.fillna(0)
```

```python
students['Package'].fillna(students['Package'].mean())
```

- Removes duplicate rows from a DataFrame.
- Returns a new DataFrame with duplicate records removed.
- Can check duplicates using all columns or selected column(s).

## Important Points
- Works mainly on a **DataFrame**.

- `subset` specifies the column(s) to consider when identifying duplicates.

- `keep=False` removes **all** duplicate rows.
- `inplace=True` modifies the original DataFrame instead of returning a new one.
- `ignore_index=True` resets the index after removing duplicates.
- Commonly used during data cleaning to eliminate repeated records.

## `apply`

## Function
`apply(func, axis=0, raw=False, result_type=None, args=(), **kwargs)`

## What it does
- Applies a function to each row or column of a Series or DataFrame.
- Used to perform custom operations that are not directly available in Pandas.
- Returns the result after applying the function.


## Important Points
- Works on both **Series** and **DataFrame**.
- For a **Series**, `apply()` applies the function to each value.
- For a **DataFrame**:
  - `axis=0` → applies function column-wise (default).
  - `axis=1` → applies function row-wise.
- Can use:
  - User-defined functions
  - Lambda functions
  - Built-in functions
- Does not modify the original DataFrame unless the result is assigned back.
- Commonly used for feature creation, data transformation, and custom logic.
- More flexible than methods like `map()` because it can work on rows and columns.

```python
import math

def sigmoid(x):
    return 1 / (1 + math.exp(-x))

# Example values
values = [-5, -2, 0, 2, 5]

for x in values:
    print(x, sigmoid(x))
```

```python
# DataFrame with two points
df = pd.DataFrame({
    "Point_A": [(4, 5), (2, 3), (1, 7)],
    "Point_B": [(9, 2), (6, 8), (4, 5)]
})

print("Original DataFrame:")
print(df)

# Calculate distance between Point_A and Point_B
df["distance"] = df.apply(
    lambda row: np.sqrt(
        (row["Point_B"][0] - row["Point_A"][0])**2 +
        (row["Point_B"][1] - row["Point_A"][1])**2
    ),
    axis=1
)

# Calculate sigmoid of distance
df["sigmoid"] = 1 / (1 + np.exp(-df["distance"]))

print("\nAfter calculation:")
print(df)
```

## `isin`

- Returns a Boolean Series/DataFrame:

## Important Points

## `corr`

## Function
`corr(method='pearson', min_periods=1, numeric_only=False)`

## What it does
- Calculates the correlation between columns of a DataFrame or between values in a Series.
- Measures the strength and direction of the relationship between numerical variables.
- Returns a correlation matrix.

## Important Points
- Works mainly on **DataFrame** numerical columns.
- Returns values between **-1 and +1**:
  - `+1` → Perfect positive correlation
  - `0` → No linear relationship
  - `-1` → Perfect negative correlation
- Default method is **Pearson correlation**.
- Other methods:
  - `pearson` → Measures linear relationship (default)
  - `spearman` → Based on rank correlation
  - `kendall` → Based on Kendall's tau correlation
- Only numerical columns are considered for correlation.
- Diagonal values are always `1` because a variable is perfectly correlated with itself.
- Commonly used in Exploratory Data Analysis (EDA) to find relationships between features.
- Helps in feature selection before machine learning models.

```python
df = pd.DataFrame({
    "CGPA": [7.5, 8.0, 8.5, 9.0, 9.5],
    "Package": [8, 10, 15, 20, 25]
})

print(df.corr())
```

## `nlargest`

## Function
`nlargest(n, columns, keep='first')`

## What it does
- Returns the rows with the **largest values** from a DataFrame or Series.
- Used to find top `n` records based on one or more columns.

## Important Points
- Works on both **Series** and **DataFrame**.
- `n` specifies the number of largest values to return.
- `columns` specifies the column(s) used for ranking (required for DataFrame).
- `keep` controls handling of duplicate values:
  - `'first'` → keeps first occurrence (default)

- Returns rows sorted in descending order.

- Commonly used for finding top performers, highest salaries, highest ratings, etc.

## `nsmallest`

## Function
`nsmallest(n, columns, keep='first')`

## What it does
- Returns the rows with the **smallest values** from a DataFrame or Series.
- Used to find bottom `n` records based on one or more columns.

## Important Points
- Works on both **Series** and **DataFrame**.
- `n` specifies the number of smallest values to return.
- `columns` specifies the column(s) used for ranking.
- `keep` controls handling of duplicate values:
  - `'first'` → keeps first occurrence (default)
  - `'last'` → keeps last occurrence
  - `False` → removes all duplicate values
- Returns rows sorted in ascending order.
- Faster than using `sort_values(ascending=True).head(n)`.
- Commonly used for finding lowest prices, lowest scores, minimum values, etc.

```python
students = pd.DataFrame({
    "Name": ["Aman", "Priya", "Rahul", "Sneha", "Karan"],
    "CGPA": [9.1, 8.7, 8.0, 9.5, 7.8],
    "Package": [25, 18, 12, 30, 10]
})

print("Original DataFrame:")
print(students)


# nlargest - Top 3 students based on Package
print("\nTop 3 Packages:")
print(students.nlargest(3, "Package"))


# nsmallest - Bottom 2 students based on CGPA
print("\nLowest 2 CGPA:")
print(students.nsmallest(2, "CGPA"))
```

## `insert`

## Function
`insert(loc, column, value, allow_duplicates=False)`

## What it does
- Adds a new column to a DataFrame at a specified position.
- Allows inserting a column between existing columns.
- Modifies the original DataFrame directly.


## Important Points
- Works only on a **DataFrame**.
- `loc` specifies the position where the new column will be inserted.
- Column indexing starts from `0`.
- `column` specifies the name of the new column.
- `value` specifies the data to be inserted.
- `insert()` modifies the original DataFrame and does not return a new DataFrame.
- By default, duplicate column names are not allowed.
- `allow_duplicates=True` allows inserting a column with an existing name.
- Different from direct assignment:
  - `df["new_col"] = value` → adds column at the end.
  - `df.insert()` → adds column at a specific position.

```python
students = pd.DataFrame({
    "Name": ["Aman", "Priya", "Rahul"],
    "CGPA": [9.1, 8.7, 8.0],
    "Package": [25, 18, 12]
})

print("Before Insert:")
print(students)

# Insert new column
students.insert(1, "Branch", ["CSE", "ECE", "ME"])

print("\nAfter Insert:")
print(students)
```

## `copy`

## Function
`copy(deep=True)`

## What it does
- Creates a copy of a Pandas Series or DataFrame.
- Allows creating an independent copy of data so changes in one object do not affect the other.
- Used to avoid modifying the original DataFrame accidentally.


## Important Points
- Works on both **Series** and **DataFrame**.
- `deep=True` creates a complete independent copy (default).
- `deep=False` creates a shallow copy where data may still share memory.
- Changes made in a deep copy do not affect the original object.
- Useful when performing data manipulation while keeping the original data safe.
- Helps avoid unexpected changes caused by assigning one DataFrame to another.
- Difference:
  - `df2 = df` → both variables point to the same DataFrame.
  - `df2 = df.copy()` → creates a separate DataFrame.

```python
students = pd.DataFrame({
    "Name": ["Aman", "Priya", "Rahul"],
    "CGPA": [9.1, 8.7, 8.0]
})

# Create copy
new_students = students.copy()

# Modify copied DataFrame
new_students["CGPA"] = [10, 9.5, 8.8]

print("Original DataFrame:")
print(students)

print("\nCopied DataFrame:")
print(new_students)
```
