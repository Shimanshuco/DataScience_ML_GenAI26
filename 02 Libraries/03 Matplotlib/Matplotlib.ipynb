# 1. 2D Line Plot (`plt.plot()`)

## What is a 2D Line Plot?

A **2D Line Plot** displays data points connected by straight lines. It is mainly used to show **trends, patterns, and changes over a continuous interval**.

## Syntax

```python
import matplotlib.pyplot as plt

plt.plot(x, y)

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Line Plot")

plt.show()
```

## When to Use a Line Plot?

Use a line plot when:

- You want to show how something changes over time.
- The x-axis represents continuous values.
- You want to identify trends or patterns.
- You want to compare multiple continuous datasets.

# 2. Matplotlib Scatter Plot (`plt.scatter()`)

## What is a Scatter Plot?

A **scatter plot** displays data points as individual dots on a graph. It is used to show the **relationship, correlation, or pattern between two numerical variables**.

Unlike a line plot, scatter plots do **not connect points with lines**.

---

## Syntax

```python
import matplotlib.pyplot as plt

plt.scatter(x, y)

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Scatter Plot")

plt.show()
```

---

# When to Use a Scatter Plot?

Use a **scatter plot** when:

- You want to find the relationship between two variables.
- You want to check correlation between data.
- You want to identify patterns or clusters.
- You want to detect outliers.
- The data points are independent of each other.

---

# Common Applications

| Field | Example |
|-------|---------|
| Business | Advertising cost vs sales |
| Finance | Risk vs return |
| Medicine | Height vs weight |
| Education | Study hours vs marks |
| Machine Learning | Feature relationship analysis |

---

# Scatter Plot Parameters

| Parameter | Purpose |
|-----------|---------|
| `color` | Changes dot color |
| `marker` | Changes dot shape |
| `s` | Changes dot size |
| `alpha` | Controls transparency |
| `label` | Adds legend |

Example:

```python
plt.scatter(
    x,
    y,
    color="red",
    marker="o",
    s=100,
    alpha=0.7
)
```

---

# Different Markers

| Marker | Code |
|--------|------|
| Circle | `"o"` |
| Square | `"s"` |
| Triangle | `"^"` |
| Star | `"*"` |
| Diamond | `"D"` |

Example:

```python
plt.scatter(x, y, marker="^")
```


# When NOT to Use Scatter Plot?

| Situation | Better Graph |
|-----------|--------------|
| Showing change over time | Line Plot |
| Comparing categories | Bar Chart |
| Showing frequency distribution | Histogram |
| Showing parts of a whole | Pie Chart |

---

# How to Identify a Scatter Plot Problem?

Ask these questions:

| Question | Use Scatter Plot? |
|----------|------------------|
| Are both variables numerical? | ✅ Yes |
| Do you want to find a relationship? | ✅ Yes |
| Are points independent? | ✅ Yes |
| Do you want to show time trend? | ❌ Use Line Plot |
| Do you want to compare categories? | ❌ Use Bar Chart |

---

# Remember

**Scatter Plot = Relationship + Correlation + Two Numerical Variables**

Examples:

- Height vs Weight
- Price vs Mileage
- Study Hours vs Marks
- Advertising Cost vs Sales

# 3. Matplotlib Bar Chart (`plt.bar()`)

## What is a Bar Chart?

A **Bar Chart** represents data using rectangular bars. The length or height of each bar shows the value of a category.

It is mainly used to **compare different categories or groups**.

---

## Syntax

```python
import matplotlib.pyplot as plt

plt.bar(x, height)

plt.xlabel("Categories")
plt.ylabel("Values")
plt.title("Bar Chart")

plt.show()
```

---

# When to Use a Bar Chart?

Use a **bar chart** when:

- You want to compare different categories.
- Your data contains discrete groups.
- You want to show differences between items.
- You want to compare quantities, counts, or values.

---

# Common Applications

| Field | Example |
|-------|---------|
| Business | Sales of different products |
| Education | Marks of different students |
| Marketing | Number of customers by month |
| Finance | Revenue comparison by company |
| Survey | Responses by category |


---

# Types of Bar Charts

## 1. Vertical Bar Chart

Bars are displayed vertically.

```python
plt.bar(category, values)
```

Example:

```
|
|       █
|   █   █
| █ █   █
-------------
 A B C
```

Used for comparing categories.

---

## 2. Horizontal Bar Chart

Bars are displayed horizontally.

```python
plt.barh(category, values)
```


Useful when category names are long.

---

# Bar Chart Parameters

| Parameter | Purpose |
|-----------|---------|
| `color` | Changes bar color |
| `width` | Changes bar thickness |
| `edgecolor` | Adds border color |
| `alpha` | Controls transparency |
| `label` | Adds legend |

Example:

```python
plt.bar(
    products,
    sales,
    color="green",
    width=0.5,
    edgecolor="black"
)
```

---

# Multiple Bar Chart

Used to compare two or more groups.


---


# When NOT to Use Bar Chart?

| Situation | Better Graph |
|-----------|--------------|
| Showing continuous change | Line Plot |
| Showing correlation | Scatter Plot |
| Showing data distribution | Histogram |
| Showing percentage parts | Pie Chart |

---

# How to Identify a Bar Chart Problem?

Ask these questions:

| Question | Use Bar Chart? |
|----------|----------------|
| Are you comparing categories? | ✅ Yes |
| Is the data grouped by names/types? | ✅ Yes |
| Are values separate for each group? | ✅ Yes |
| Is the data changing continuously? | ❌ Use Line Plot |
| Are you finding relationships? | ❌ Use Scatter Plot |

---

# Remember

**Bar Chart = Compare Categories**

Examples:

- Product sales comparison
- Student marks comparison
- Population by country
- Revenue by company
- Car sales by brand

# 4. Matplotlib Histogram (`plt.hist()`)

## What is a Histogram?

A **Histogram** is a graph used to represent the **frequency distribution of numerical data**.

It divides data into intervals called **bins** and shows how many values fall into each interval.

A histogram helps us understand:
- Data distribution
- Frequency of values
- Spread of data
- Skewness
- Outliers

---

## Syntax

```python
import matplotlib.pyplot as plt

plt.hist(data, bins)

plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Histogram")

plt.show()
```

---

# When to Use a Histogram?

Use a **histogram** when:

- You have continuous numerical data.
- You want to know how data is distributed.
- You want to count how many values fall in a range.
- You want to identify the shape of data.
- You want to detect unusual values (outliers).

---

# Examples of Histogram Use

| Field | Example |
|-------|---------|
| Education | Distribution of student marks |
| Healthcare | Patient age distribution |
| Finance | Income distribution |
| Business | Customer purchase amounts |
| Data Science | Feature distribution before ML modeling |

---


# Understanding Bins

Bins are the intervals used to group data.

Example:

Data:

```
10, 15, 18, 25, 30, 35, 40
```

With 3 bins:

```
10 - 20  → 3 values
20 - 30  → 1 value
30 - 40  → 3 values
```

The histogram displays these counts.

---

# Important Parameters

| Parameter | Description |
|-----------|-------------|
| `bins` | Number of intervals |
| `color` | Changes bar color |
| `edgecolor` | Adds border around bars |
| `alpha` | Controls transparency |
| `density` | Displays probability density |

Example:

```python
plt.hist(
    data,
    bins=10,
    color="blue",
    edgecolor="black",
    alpha=0.7
)
```

---

# Types of Data Suitable for Histogram

| Suitable Data | Example |
|--------------|---------|
| Continuous numerical data | Height of people |
| Measurements | Temperature readings |
| Scores | Exam marks |
| Financial data | Salary distribution |

---


# When NOT to Use Histogram?

| Situation | Use Instead |
|-----------|-------------|
| Comparing categories | Bar Chart |
| Showing trend over time | Line Plot |
| Showing relationship between variables | Scatter Plot |
| Showing percentage contribution | Pie Chart |

---

# How to Identify a Histogram Problem?

Ask:

| Question | Answer |
|----------|--------|
| Is the data numerical? | ✅ Use Histogram |
| Do you want frequency/count? | ✅ Use Histogram |
| Do you want data distribution? | ✅ Use Histogram |
| Are you comparing categories? | ❌ Use Bar Chart |
| Are you finding correlation? | ❌ Use Scatter Plot |

---

# Remember

**Histogram = Distribution + Frequency + Continuous Numerical Data**

Examples:

- Student marks distribution
- Age distribution
- Height distribution
- Salary distribution
- Customer spending distribution

````python
data = [32,45,56,10,15,27,61]

plt.hist(data,bins=[10,25,40,55,70])
````

# 5. Matplotlib Pie Chart (`plt.pie()`)

## What is a Pie Chart?

A **Pie Chart** represents data as slices of a circle. Each slice shows the **proportion or percentage contribution** of a category compared to the whole.

The complete circle represents **100% of the data**.

---

## Syntax

```python
import matplotlib.pyplot as plt

plt.pie(values, labels=categories)

plt.title("Pie Chart")

plt.show()
```

---

# When to Use a Pie Chart?

Use a **pie chart** when:

- You want to show how parts make up a whole.
- You want to display percentages or proportions.
- You have a small number of categories.
- You want to compare shares of a total.

---

# Common Applications

| Field | Example |
|-------|---------|
| Business | Market share of companies |
| Finance | Expense distribution |
| Marketing | Customer segments |
| Survey | Percentage of responses |
| Education | Student grade distribution |


---

# Important Parameters in `plt.pie()`

| Parameter | Purpose |
|-----------|---------|
| `labels` | Names of categories |
| `autopct` | Displays percentage values |
| `colors` | Changes slice colors |
| `explode` | Separates a slice from the pie |
| `startangle` | Rotates the chart |
| `shadow` | Adds shadow effect |

---


# Pie Chart Rules

| Rule | Explanation |
|------|-------------|
| Use fewer categories | Too many slices make it confusing |
| Total should represent 100% | Parts must make a complete whole |
| Categories should be meaningful | Each slice should have a clear purpose |
| Avoid very similar values | Hard to compare similar-sized slices |

---


# When NOT to Use Pie Chart?

| Situation | Better Graph |
|-----------|--------------|
| Comparing many categories | Bar Chart |
| Showing changes over time | Line Plot |
| Showing relationships | Scatter Plot |
| Showing data distribution | Histogram |

---

# How to Identify a Pie Chart Problem?

Ask these questions:

| Question | Use Pie Chart? |
|----------|----------------|
| Does the data represent parts of a whole? | ✅ Yes |
| Do values add up to 100%? | ✅ Yes |
| Are there only a few categories? | ✅ Yes |
| Do you want to show changes over time? | ❌ Use Line Plot |
| Do you want exact value comparison? | ❌ Use Bar Chart |

---

# Remember

**Pie Chart = Percentage + Parts of a Whole**

Examples:

- Budget distribution
- Market share
- Population percentage
- Survey results
- Expense breakdown

````python
data = [23,45,100,20,49]
subjects = ['eng','science','maths','sst','hindi']
plt.pie(data,labels=subjects)

plt.show()
````
