### Why NumPy is Fastest

# Vectorization

- Vectorization is the process of performing operations on entire NumPy arrays without using explicit loops or indexing in Python. The operations are executed internally using optimized C code.

- Vectorized code is more concise and easier to read, requiring fewer lines of code compared to traditional looping methods.

- Using fewer lines of code reduces the chances of errors and bugs, making programs more reliable.

- Vectorization makes code resemble standard mathematical notation, making mathematical computations easier to understand and implement.

- It produces more Pythonic and efficient code by avoiding slow and difficult-to-read `for` loops.

# Broadcasting

- Broadcasting is the mechanism that allows NumPy to perform implicit element-by-element operations on arrays.

- Broadcasting applies to arithmetic, logical, bitwise, and functional operations, enabling efficient computation on arrays.

- It works when arrays have the same shape, when one operand is a scalar, or when arrays have compatible shapes.

- If a smaller array can be expanded to match the shape of a larger array without ambiguity, NumPy automatically performs the operation using broadcasting.

## Creating NumPy Arrays

```python
# Importing np.array
import numpy as np
```

```python
# dtype
# Specifies the data type of elements in a NumPy array.

# Syntax:
# np.array(data, dtype=type)

# Important:
# dtype controls how data is stored in memory.
# All elements in a NumPy array generally have the same dtype.
# Changing dtype can improve memory efficiency or precision.

# Common dtypes:
# int, float, bool, complex

c = np.array([1,2,0,1,-1],dtype=float)
print(c)
c = np.array([1,2,0,1,-1],dtype=bool)
print(c)
```

```python
# np.arange()
# Creates a NumPy array with evenly spaced values within a given range.

# Syntax:
# np.arange(start, stop, step)

# Parameters:
# start -> starting value (included)
# stop  -> ending value (excluded)
# step  -> difference between each value

# Important:
# The stop value is NOT included.
# Works similar to Python range() but returns a NumPy array.
# Useful for generating sequences of numbers.

d = np.arange(1,11) # 1 to 10 
print(d)

d = np.arange(2,21,2) # Table of 2
print(d)
```

```python
# reshape()
# Used to change the shape (dimensions) of a NumPy array.
# It does not change the data, only the arrangement of elements.

# Syntax:
# array.reshape(rows, columns)

# Important:
# The total number of elements must remain the same.
# rows × columns = total elements in the array.
# Otherwise, NumPy raises a ValueError.

e = np.arange(1,11).reshape(2,5)
print(e)
```

```python
# np.ones()
# Creates a NumPy array filled with 1s.

# Syntax:
# np.ones(shape, dtype=data_type)

# Parameters:
# shape -> Size of the array (e.g., 5, (2,3), (3,3))
# dtype -> Optional, specifies the data type

# Important:
# By default, elements are of float type (1.0).
# Useful for initializing arrays with all values set to 1.

f = np.ones((2, 3))
print(f)
```

```python
# np.zeros()
# Creates a NumPy array filled with 0s.

# Syntax:
# np.zeros(shape, dtype=data_type)

# Parameters:
# shape -> size of the array (e.g., 5, (2,3), (4,4))
# dtype -> optional, defines the data type

# Important:
# By default, values are float (0.0).
# Used for initializing arrays before computation.

g = np.zeros((4,3))
print(g)
```

```python
# np.random
# Used to generate random numbers in NumPy.

# Common functions:

# np.random.rand()
# Generates random float numbers between 0 and 1.
# Shape is given as separate arguments
arr1 = np.random.rand(3)
print(arr1)

# np.random.randint(start, stop, size)
# Generates random integers from start to stop (stop excluded).
arr2 = np.random.randint(1, 10, 5)
print(arr2)

# np.random.randn()
# Generates numbers from a normal (Gaussian) distribution.
arr3 = np.random.randn(2, 2)
print(arr3)

# np.random.random()
# Shape is given as a tuple
arr4 = np.random.random((2, 3))
print(arr4)

# Important:
# Useful in simulations, testing, machine learning, and data generation.
# Numbers are pseudo-random (generated using an algorithm, not truly random).
```

```python
# np.linspace()
# Generates evenly spaced numbers over a specified range.

# Syntax:
# np.linspace(start, stop, num)

# Parameters:
# start -> starting value (included)
# stop  -> ending value (included)
# num   -> number of values to generate

# Important:
# Unlike arange(), the stop value IS included.
# It does not use step size; instead it uses number of points.
# Useful for graphs and plotting smooth curves.

h = np.linspace(-10,10,10)
print(h)
```

```python
# np.identity()
# Creates an identity matrix (square matrix with 1s on diagonal and 0s elsewhere).

# Syntax:
# np.identity(n, dtype=float)

# Parameters:
# n -> size of the matrix (n x n)
# dtype -> optional data type (default is float)

# Important:
# Only square matrices are created (rows = columns).
# Diagonal elements are 1, all others are 0.
# Used in linear algebra.

# Example:
i = np.identity(3)
print(i)
```

## Indexing and Slicing

```python
a1 = np.arange(10)
print("a1:", a1)

a2 = np.arange(12).reshape(3,4)
print("\na2:\n", a2)

a3 = np.arange(8).reshape(2,2,2)
print("\na3:\n", a3)
```

```python
# -------------------------
# 1D Array (a1) Indexing & Slicing
# -------------------------
print("\n1D Indexing:")
print(a1[0])      # first element
print(a1[-1])     # last element

print("\n1D Slicing:")
print(a1[2:7])    # from index 2 to 6
print(a1[:5])     # first 5 elements
print(a1[::2])    # step = 2

# -------------------------
# 2D Array (a2) Indexing & Slicing
# -------------------------
print("\n2D Indexing:")
print(a2[0, 0])   # first row first column
print(a2[2, 3])   # last element

print("\n2D Slicing:")
print(a2[0])      # first row
print(a2[:, 1])   # all rows, column 1
print(a2[1:, 2:]) # sub-matrix

# -------------------------
# 3D Array (a3) Indexing & Slicing
# -------------------------
print("\n3D Indexing:")
print(a3[0])        # first 2D block
print(a3[1, 0, 1])  # specific element

print("\n3D Slicing:")
print(a3[:, 1])     # second row of each block
print(a3[:, :, 1])  # second column of all blocks
```

## Iteration

```python
a1 = np.arange(10)
print("a1:\n", a1)

a2 = np.arange(12).reshape(3,4)
print("\na2:\n", a2)

a3 = np.arange(8).reshape(2,2,2)
print("\na3:\n", a3)

# -------------------------
# Iteration in 1D array
# -------------------------
print("\n1D iteration:")
for i in a1:
    print(i)

# -------------------------
# Iteration in 2D array (row-wise)
# -------------------------
print("\n2D iteration (rows):")
for row in a2:
    print(row)

print("\n2D iteration (element-wise):")
for i in np.nditer(a2):
    print(i)

# -------------------------
# Iteration in 3D array
# -------------------------
print("\n3D iteration (blocks):")
for block in a3:
    print(block)

print("\n3D full iteration (all elements):")
for i in np.nditer(a3):
    print(i)
```

```python
a = np.arange(12).reshape(3,4)
print("Original array:\n", a)

# -------------------------
# 1) reshape()
# -------------------------
# Changes shape without changing data
print("\nReshape (2,6):\n", a.reshape(2,6))

# -------------------------
# 2) transpose()
# -------------------------
# Swaps rows and columns (for 2D arrays)
print("\nTranspose:\n", a.transpose())
# OR
print("\nShortcut transpose:\n", a.T)

# -------------------------
# 3) ravel()
# -------------------------
# Converts multi-D array into 1D array (flatten view)
print("\nRavel:\n", a.ravel())
```

## Stacking

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# 2D arrays for better understanding
a2 = np.array([[1, 2, 3],
               [4, 5, 6]])

b2 = np.array([[7, 8, 9],
               [10, 11, 12]])

print("a2:\n", a2)
print("b2:\n", b2)

# -------------------------
# 1) Horizontal stacking  : Joins arrays side by side (columns increase)
# -------------------------
print("\nHorizontal stacking (column-wise):")
print(np.hstack((a2, b2)))

# -------------------------
# 2) Vertical stacking  : Joins arrays one above another (rows increase)
# -------------------------
print("\nVertical stacking (row-wise):")
print(np.vstack((a2, b2)))
```

## Spliting

```python
a = np.arange(12)
print("Original array:\n", a)

# -------------------------
# 1) 1D splitting
# -------------------------
print("\n1D split:")
print(np.split(a, 3))   # split into 3 equal parts

# -------------------------
# 2) 2D array splitting
# -------------------------
a2 = np.arange(12).reshape(3,4)
print("\n2D array:\n", a2)

print("\nHorizontal split (column-wise):")
print(np.hsplit(a2, 2))   # split columns

print("\nVertical split (row-wise):")
print(np.vsplit(a2, 3))   # split rows

# -------------------------
# 3) uneven split (example)
# -------------------------
a3 = np.arange(10)
print("\nEven split:")
print(np.array_split(a3, 3))  # works even if not perfectly divisible
```
