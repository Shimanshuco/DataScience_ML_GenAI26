# np.sort

## 1) Name
**np.sort** – NumPy function used to sort the elements of an array.

## 2) Syntax
```python
np.sort(a, axis=-1, kind=None, order=None)
```

## 3) Usage
Sorts array elements in ascending order and returns a new sorted array.

## Note
- Returns a new sorted array.
- Does not modify the original array.

````python
import numpy as np
````

````python
a = np.random.randint(1,100,25) # 1-D
a
````

````python
print(np.sort(a))
````

````python
b = np.arange(1, 21).reshape(4, 5)
print(b)
````

````python
print(np.sort(b)) # row wise sorting
````

````python
print(np.sort(b , axis = 0 )) # Col wise sorting
````

````python
# Descending Order Sorting
print(np.sort(a)[::-1])
````

# np.append

## 1) Name
**np.append** – NumPy function used to add values to the end of an array.

## 2) Syntax
```python
np.append(arr, values, axis=None)
```

### Parameters
- `arr` : Original array.
- `values` : Values to append.
- `axis` : Axis along which values are appended.

## 3) Usage
Used to append elements to an existing array. It returns a new array and does not modify the original array.

# np.concatenate

## 1) Name
**np.concatenate** – NumPy function used to join two or more arrays along an existing axis.

## 2) Syntax
```python
np.concatenate((arr1, arr2, ...), axis=0)
```

### Parameters
- `arr1, arr2, ...` : Arrays to join.
- `axis` : Axis along which the arrays are joined.
  - `axis=0` → Join rows (vertical).
  - `axis=1` → Join columns (horizontal).

## 3) Usage
Used to combine multiple arrays into a single array.

## Note
- Arrays must have compatible dimensions.
- For `axis=0`, the number of columns must match.
- For `axis=1`, the number of rows must match.
- Returns a new array; the original arrays are unchanged.

````python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

result = np.concatenate((a, b))

print(result)
````

````python
# Row (axis = 0)
a = np.array([[1, 2],  
              [3, 4]])

b = np.array([[5, 6]])

result = np.concatenate((a, b), axis=0)

print(result)
````

````python
#  Col (axis = 1)
a = np.array([[1, 2],
              [3, 4]])

b = np.array([[5],
              [6]])

result = np.concatenate((a, b), axis=1)

print(result)
````

# np.unique

## 1) Name
**np.unique** – NumPy function used to find the unique (distinct) elements in an array.

## 2) Syntax
```python
np.unique(ar, return_index=False, return_inverse=False,
          return_counts=False, axis=None)
```

### Parameters
- `ar` : Input array.
- `return_index` : Returns indices of unique values.
- `return_inverse` : Returns indices to reconstruct the original array.
- `return_counts` : Returns the count of each unique value.
- `axis` : Find unique values along a specified axis.

## 3) Usage
Used to remove duplicate values and return the sorted unique elements.



## Note
- Returns unique values in sorted order.
- Removes duplicate elements automatically.
- Can also return counts, indices, and inverse mappings.

````python
arr = np.array([1, 2, 2, 3, 3, 3, 4])

result = np.unique(arr)

print(result)
````

````python
arr = np.array([1, 2, 2, 3, 3, 3, 4,3])

unique_values, counts = np.unique(arr, return_counts=True)

print(unique_values)
print(counts)
````

````python
arr = np.array([[1, 2],
                [1, 2],
                [3, 4]])

result = np.unique(arr, axis=0)

print(result)
````

# np.where

## 1) Name
**np.where** – NumPy function used to find indices based on a condition or to choose values conditionally.

---

## 2) Syntax

### Case 1: Condition only
```python
np.where(condition)
```

### Case 2: Choose between two values
```python
np.where(condition, x, y)
```

### Parameters
- `condition` : Condition to check (True/False).
- `x` : Value used where condition is True.
- `y` : Value used where condition is False.

---

## 3) Usage
Used for:
- Finding positions of elements that satisfy a condition
- Conditional replacement in arrays



## Note
- Returns indices when only condition is given.
- Works like a vectorized `if-else`.
- Very useful for fast filtering and data transformation.

````python
arr = np.array([10, 20, 30, 40, 50])

result = np.where(arr > 25)

print(result)
````

````python
arr = np.array([10, 20, 30, 40])

result = np.where(arr > 25, "High", "Low")

print(result)
````

````python
arr = np.array([1, 2, 3, 4, 5])

result = np.where(arr % 2 == 0, 0, 1)

print(result)
````

# np.corrcoef

## 1) Name
**np.corrcoef** – NumPy function used to compute the correlation coefficient between variables.

---

## 2) Syntax
```python
np.corrcoef(x, y=None, rowvar=True)
```

### Parameters
- `x` : Input array(s).
- `y` : Optional second variable.
- `rowvar` :
  - `True` (default): each row is a variable.
  - `False`: each column is a variable.

---

## 3) Usage
Used to measure the **relationship (correlation)** between two or more variables.
- Value ranges from **-1 to +1**
  - `+1` → perfect positive correlation  
  - `0` → no correlation  
  - `-1` → perfect negative correlation  



## Note
- Produces a **correlation matrix**.
- Helps in statistics, data analysis, and machine learning.
- Shows strength and direction of linear relationships.

````python
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

result = np.corrcoef(x, y)

print(result)
````

````python
data = np.array([
    [1, 2, 3, 4, 5],
    [2, 4, 6, 8, 10],
    [5, 4, 3, 2, 1]
])

result = np.corrcoef(data)

print(result)
````

````python
data = np.array([
    [1, 2, 3, 4, 5],
    [2, 4, 6, 8, 10],
    [5, 4, 3, 2, 1]
])

result = np.corrcoef(data)

print(result)
````

# np.isin

## 1) Name
**np.isin** – NumPy function used to check whether elements of one array exist in another array.

---

## 2) Syntax
```python
np.isin(element, test_elements, assume_unique=False, invert=False)
```

### Parameters
- `element` : Input array to check.
- `test_elements` : Values to compare against.
- `assume_unique` : If True, speeds up processing (assumes inputs are unique).
- `invert` : If True, returns elements NOT in `test_elements`.

---

## 3) Usage
Used to perform **membership testing** (like `in` keyword, but for arrays).



## Note
- Returns a **boolean array**.
- Very useful for filtering data.
- Similar to SQL `IN` condition or Python `in`, but works on full arrays efficiently.

````python
arr = np.array([1, 2, 3, 4, 5])

result = np.isin(arr, [2, 4, 6])

print(result)
````

````python
arr = np.array([10, 20, 30, 40, 50])

result = arr[np.isin(arr, [20, 50])]

print(result)
````

````python
arr = np.array([1, 2, 3, 4, 5])

result = np.isin(arr, [2, 4], invert=True)

print(result)
````

- `invert` : If True, returns elements NOT in `ar2`.

---

## 3) Usage
Used for **membership testing** (like `np.isin`, but older version):
👉 Checks which elements of `ar1` exist in `ar2`.


## Note
- Returns a **boolean 1D array**.
- Similar to `np.isin`, but `np.isin` is preferred in modern NumPy.
- Useful for filtering and membership checks.

````python
a = np.array([1, 2, 3, 4, 5])
b = np.array([2, 4, 6])

result = np.in1d(a, b)

print(result)
````

````python
a = np.array([10, 20, 30, 40, 50])
b = np.array([20, 50])

result = a[np.in1d(a, b)]

print(result)
````

````python
a = np.array([1, 2, 3, 4, 5])
b = np.array([2, 4])

result = np.in1d(a, b, invert=True)

print(result)
````

# np.clip()

## 1) Name
**np.clip** – NumPy function used to limit (restrict) values in an array within a specified range.

---

## 2) Syntax
```python
np.clip(a, a_min, a_max)
```

### Parameters
- `a` : Input array.
- `a_min` : Minimum allowed value.
- `a_max` : Maximum allowed value.

---

## 3) Usage
Used to **bound values** so they do not go below or above a given range.

👉 Values smaller than `a_min` become `a_min`  
👉 Values greater than `a_max` become `a_max`


## Note
- Very useful in **image processing** (pixel range 0–255).
- Also used in ML to prevent extreme values.
- Does not change array shape, only values.

````python
arr = np.array([1, 5, 10, 15, 20])

result = np.clip(arr, 5, 15)

print(result)
````

````python
arr = np.array([-10, -5, 0, 5, 10])

result = np.clip(arr, 0, 8)

print(result)
````

````python
arr = np.array([50, 120, 200, 300])

result = np.clip(arr, 0, 255)

print(result)
````
